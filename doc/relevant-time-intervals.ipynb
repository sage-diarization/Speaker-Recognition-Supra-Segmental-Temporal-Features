Prosody is the part of speech that lies above the individual segments (consonants and vowels). It is about how a stretch of speech is said: pitch, loudness, timing and rhythm, rather than which sounds are said. Phonetics often uses "prosodic" and suprasegmental as near-synonyms, after Lehiste's (1970) book Suprasegmentals. Some authors limit "prosody" to the features that serve a linguistic function.

Units it operates on

These are the units of the prosodic hierarchy (Selkirk 1984; Nespor & Vogel 1986): syllable → foot → prosodic word → phonological phrase → intonational phrase → utterance. Above the phone level, these are the same units as in the time-span table from the research earlier in this conversation.

Functions

- Linguistic: word stress, lexical tone (as in Mandarin), sentence focus and emphasis, phrasing a rising tune for questions).
- Paralinguistic: emotion, attitude, emphasis.
- Extralinguistic: speaker identity, age, dialect. This is the function behind the supra-segmenton in Neururer et al.
